```python
import random
  #rock = 0
  #paper = 1
  #scissors =2
def game():
    print("\nrock = 0   👊",end="")
    print("     paper = 1   🖐️",end="")
    print("     scissors = 2   ✌")

    computer = random.choice(['0','1','2'])
    print("\n===============================")
    print("Rock Paper Scissors Shoot!!")
    print("===============================\n")
    player=input("chose your sign : ")

    if (computer == '0' and player =='1'):
      print("\n==================================================")
      print("  computer = rock 👊          player = paper 🖐️")

      print("\n\t\tplayer won 🖐️")
      print("==================================================")
    elif (computer == '0' and player =='2'):
      print("\n==================================================")
      print("\ncomputer = rock 👊          player = scissors ✌")

      print("\n\t\tomputer won 👊")
      print("==================================================")
    elif (computer == '0' and player =='0'):
      print("\n==================================================")
      print("\ncomputer = rock 👊           player = rock 👊")

      print("\n\t\tDraw !!!")
      print("==================================================")
    elif (computer == '1' and player == '0'):
      print("\n==================================================")
      print("\ncomputer = paper 🖐️          player = rock 👊")

      print("\n\t\tcomputer won 🖐️")
      print("==================================================")
    elif (computer == '1' and player == '1'):
      print("\n==================================================")
      print("\ncomputer = paper 🖐️           player = paper 🖐️")

      print("\n\t\t Draw !!!")
      print("==================================================")
    elif (computer == '1' and player =='2'):
      print("\n==================================================")
      print("\ncomputer = paper 🖐️          player = scissors ✌")

      print("\n\t\tplayer won ✌ !!!")
      print("==================================================")
    elif (computer == '2' and player == '0'):
      print("\n==================================================")
      print("\ncomputer = scissors ✌         player = rock 👊")

      print("\n\t\tplayer won 👊 !!!")
      print("==================================================")
    elif (computer == '2' and player == '1'):
      print("\n==================================================")
      print("\ncomputer = scissors ✌          player = paper 🖐️ ")

      print("\n\t\tcomputer won ✌ ")
      print("==================================================")
    elif (computer == '2' and player =='2'):
      print("\n==================================================")
      print("\ncomputer = scissors ✌          player = scissors ✌")

      print("\n\t\t\Draw !!!")
      print("==================================================")
    else:
      print("\n===================")
      print("  invalid input !!!")
      print("===================\n")


for i in range(3):
  game()

```
